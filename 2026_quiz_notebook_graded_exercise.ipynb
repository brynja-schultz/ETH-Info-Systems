### References between tables
- Run the following query

You don't have to understand the query but you should interpret the output correctly.

````python
%%sql
SELECT
    tc.table_name AS foreign_table, 
    kcu.column_name AS foreign_column, 
    ccu.table_name AS referenced_table,
    ccu.column_name AS referenced_column 
FROM 
    information_schema.table_constraints AS tc 
    JOIN information_schema.key_column_usage AS kcu
      ON tc.constraint_name = kcu.constraint_name
      AND tc.table_schema = kcu.table_schema
    JOIN information_schema.constraint_column_usage AS ccu
      ON ccu.constraint_name = tc.constraint_name
      AND ccu.table_schema = tc.table_schema
WHERE tc.constraint_type = 'FOREIGN KEY'
AND (tc.table_name = 'ranking' OR tc.table_name = 'matches');
````

### Extended Projection
For the following tasks, you will need to use the **extended projection** operator. This operator uses the same symbol as the normal projection, and allows you to create new attributes, by performing operations on the existing attributes of the relation. For example, assume that you have the following relation, with the name **numbers** and two attributes **Aa** and **Bb** :  

|Aa   |Bb   |
|---:|---:|
|   5|   3|
|   2|   1|
|   4|   7|  

If for example you wanted to create a new relation with the result of subtracting the numbers in column **Bb** from those in column **Aa** , then you would write this in relational algebra as:  

$$ \text{subtraction} = \pi_{(\text{Aa} - \text{Bb}) \rightarrow \text{result}}(\text{numbers}) $$

Executing this operation would result in the following **subtraction** table:  

|result|
|-----:|
|     2|
|     1|
|    -3|  

The __extended projection__ operator can be used together with the normal projection operation at the same time. Consider the following operation:  

$$ \text{subtraction} = \pi_{\text{Aa}, \text{Bb}, (\text{Aa} - \text{Bb}) \rightarrow \text{result}}(\text{numbers}) $$

The result is the following __subtraction__ table:  

|Aa   |Bb   |result|
|---:|---:|-----:|
|   5|   3|     2|
|   2|   1|     1|
|   4|   7|    -3|  


In SQL, the __extended projection__ operator works exactly the same way. The above operation can be written as (assuming the __numbers__ table has been already defined):  

``` sql
SELECT Aa, Bb, (Aa - Bb) AS result
FROM numbers;
```

### Selection on grouped rows: HAVING
For the following tasks, you will also need to use the **selection** operator **on grouped rows**. This operator uses the same symbol as the normal selection and performs a filter operation based on aggregated results in the relation. For example, assume that you have the following relation, with the name **numbers** and two attributes **Aa** and **Bb** :  

|Aa   |Bb   |
|---:|---:|
|   5|   3|
|   4|   1|
|   4|   7|  

If for example you wanted to create a new relation with the result of grouping by the column **Aa** and the corresponding sum of the integers of column **Bb**, then you would write this in relational algebra as:  

$$ \gamma_{\text{Aa}, \text{ SUM(Bb)} \rightarrow \text{sum}}(\text{numbers}) $$

Executing this operation would result in the following __"grouping"__ table:  

|Aa   |sum   |
|---:|---:|
|   5|   3|
|   4|   8| 

Further you can select only rows, which satisfy certain conditions on a aggregated value. Considering the following operation, you would get only rows where the sum of integers of column **Bb** are greater than 7:  

$$ \sigma_{\text{sum} > 7}\big(\gamma_{\text{Aa}, \text{ SUM(Bb)} \rightarrow \text{sum}}(\text{numbers})\big) $$

The result is the following __"selection on grouped rows"__ table:  

|Aa   |sum   |
|---:|---:|
|   4|   8| 

In SQL, the **HAVING** _(selection on grouped rows)_ operator is different to the **WHERE** *(normal selection)*. **HAVING** is generally used together with **GROUP BY** and cannot be simply substituted with **WHERE**. For further details, you can have a look at the [PostgreSQL documentation](https://www.postgresql.org/docs/current/tutorial-agg.html).

The above operation **can** be written as (assuming the __numbers__ table has been already defined):  

``` sql
SELECT Aa, SUM(Bb) AS sum
FROM numbers
GROUP BY Aa
HAVING SUM(Bb) > 7;
``` 

The above operation **cannot** be written as:

``` sql
SELECT Aa, SUM(Bb) AS sum
FROM numbers
GROUP BY Aa
WHERE SUM(Bb) > 7;
``` 
or

``` sql
SELECT Aa, SUM(Bb) AS sum
FROM numbers
WHERE SUM(Bb) > 7
GROUP BY Aa;
```

### Task 1

Which player spent the **most time (in hours)** on court across matches they **lost**?

In the answer, please provide the first and last name of the player and their total time on court (**in hours**) across matches they **lost**, rounded to two decimal places.<br>

**Tip:** To round a number to a specific number of decimal places, you can use the [ROUND](https://www.postgresql.org/docs/current/functions-math.html#FUNCTIONS-MATH-FUNC-TABLE) function.

````python
%%sql
SELECT players.first_name, players.last_name, ROUND(SUM(matches.minutes) / 60.0, 2)
FROM ise_quiz.players AS players
JOIN ise_quiz.matches AS matches 
    ON players.player_id = matches.player_id
WHERE outcome = 0
GROUP BY players.player_id
ORDER BY SUM(matches.minutes) DESC
LIMIT 1;

````

### Task 2

A match is called *long* if it lasted longer than 120 minutes. Which player has the **highest** fraction of *long* matches among their own matches? <br>

In the answer, please provide the first and last name of the player, together with the fraction of matches that were *long*, rounded to two decimal places and expressed as a value between 0 and 1.<br>

<details>
<summary><b>Tip 1</b> (click to reveal)</summary>

To calculate an average/percentage of integer values, you might want to look up PostgreSQL's [CAST operator](https://www.postgresql.org/docs/current/sql-expressions.html#SQL-SYNTAX-TYPE-CASTS); alternatively you can also use `(expression)::type` to type-cast (which is needed to use rational-number division).
</details>
    
<details>
<summary><b>Tip 2</b> (click to reveal)</summary>
    
We can use the `CASE` expression from SQL for conditional summation. [This article](https://www.geeksforgeeks.org/postgresql/conditional-summation-in-postgresql/) might be helpful ;)<br>
You can also consult PostgreSQL's [CASE documentation](https://www.postgresql.org/docs/current/functions-conditional.html#FUNCTIONS-CASE).
</details>

````python
%%sql   
SELECT players.first_name, players.last_name, ROUND(SUM(CASE WHEN matches.minutes > 120 THEN 1.0 ELSE 0.0 END) / COUNT(matches.player_id), 2) AS long_matches
FROM ise_quiz.players AS players
JOIN ise_quiz.matches AS matches 
    ON players.player_id = matches.player_id
GROUP BY players.player_id
ORDER BY long_matches DESC
LIMIT 1;
````

### Task 3

Considering every tournament in the `tournaments` table, in which tournament did **Novak Djokovic (player_id = 4920)** play the **fewest** matches?

In the answer, please provide the name of the tournament where he played the fewest matches and the number of matches.
<br>

<details>
<summary><b>Tip 1</b> (click to reveal)</summary>

</details>

<details>
<summary><b>Tip 2</b> (click to reveal)</summary>

````python
%%sql
SELECT tours.name, COUNT(matches.player_id) AS num_plays
FROM ise_quiz.tournaments AS tours 
LEFT OUTER JOIN(
        SELECT *
        FROM ise_quiz.matches 
        WHERE player_id = 4920
    ) AS matches
    ON tours.tournament_id = matches.tournament_id
GROUP BY tours.tournament_id
ORDER BY num_plays ASC
LIMIT 1;
````

### Task 4

Across all player_id and pairs of consecutive years, which player_id had the **largest** improvement in ranking? A smaller rank number means a better ranking, and there is one ranking record per player per year.

In the answer, please provide the player_id, the two years and the improvement in ranking, expressed as a positive integer. For example, moving from rank 10 to rank 5 means an improvement of 5 positions.

<details>
<summary><b>Tip 1</b> (click to reveal)</summary>
    
To get the year of a date, you can use the [EXTRACT](https://www.postgresql.org/docs/current/functions-datetime.html) function.
</details>

<details>
<summary><b>Tip 2</b> (click to reveal)</summary>

You can join a table to itself, using different aliases for the two copies. How could you use this to compare two ranking records for the same player?
</details>
                                                                                                                 
<details>
<summary><b>Tip 3</b> (click to reveal)</summary>

Check that your paired records are consecutive years and remember that a smaller rank number means a better ranking.
</details>

````python
%%sql
SELECT p1.player_id, EXTRACT(ISOYEAR FROM p1.rank_date), EXTRACT(ISOYEAR FROM p2.rank_date),  (p1.rank - p2.rank) AS improvement
FROM ise_quiz.ranking AS p1
JOIN ise_quiz.ranking AS p2 
    ON EXTRACT(ISOYEAR FROM p1.rank_date) = (EXTRACT(ISOYEAR FROM p2.rank_date) - 1)
    AND p1.player_id = p2.player_id
ORDER BY improvement DESC
LIMIT 1;
````

### Task 5

For each tournament, compute every player’s share of the tournament’s total recorded minutes (player’s minutes at that tournament / total minutes recorded at that tournament). Among player-tournament combinations where this share is **greater than 40%**, which combination has the **smallest** share?

In the answer, please provide the tournament name, the player_id and the share for that combination as a number between 0 and 1, rounded to three decimal places.

<details>
<summary><b>Tip 1</b> (click to reveal)</summary>

It may help to calculate the tournament totals separately and make them available when calculating each player’s share. A nested query or a `WITH` clause can help organize this.
</details>

````python
%%sql
WITH tournament_totals AS(
    SELECT tournament_id, SUM(minutes) AS total
    FROM ise_quiz.matches
    GROUP BY tournament_id
)
SELECT t.name, m.player_id, ROUND(SUM(m.minutes)::numeric / tm.total, 3) AS share
FROM ise_quiz.matches AS m
JOIN tournament_totals AS tm
    ON tm.tournament_id = m.tournament_id
JOIN ise_quiz.tournaments as t
    ON m.tournament_id = t.tournament_id
GROUP BY t.tournament_id, m.player_id, tm.total
HAVING (SUM(m.minutes)::numeric / tm.total) > 0.40
ORDER BY share ASC
LIMIT 1; 

````
